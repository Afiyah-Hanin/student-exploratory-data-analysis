```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Student Exploratory Data Analysis")

# Sample student data
data = {
    "Name": ["Aisha", "Rahul", "Priya", "Arjun", "Sara"],
    "Age": [19, 20, 19, 21, 20],
    "Marks": [85, 72, 91, 68, 88],
    "Attendance": [90, 78, 95, 70, 92]
}

df = pd.DataFrame(data)

# Display data
print(df)

# Basic information
print("\nDataset Information:")
print(df.info())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Average marks
print("\nAverage Marks:", df["Marks"].mean())

# Average attendance
print("Average Attendance:", df["Attendance"].mean())

# Marks visualization
plt.bar(df["Name"], df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks Analysis")
plt.show()
```

```python
# Student Exploratory Data Analysis - Further Analysis

# 1. Calculate average marks and attendance
average_marks = df["Marks"].mean()
average_attendance = df["Attendance"].mean()

print("===== STUDENT ANALYSIS =====")
print("Average Marks:", round(average_marks, 2))
print("Average Attendance:", round(average_attendance, 2), "%")


# 2. Find the student with the highest marks
highest_marks = df.loc[df["Marks"].idxmax()]

print("\n===== HIGHEST MARKS =====")
print("Student:", highest_marks["Name"])
print("Marks:", highest_marks["Marks"])


# 3. Find the student with the lowest marks
lowest_marks = df.loc[df["Marks"].idxmin()]

print("\n===== LOWEST MARKS =====")
print("Student:", lowest_marks["Name"])
print("Marks:", lowest_marks["Marks"])


# 4. Find the student with the highest attendance
highest_attendance = df.loc[df["Attendance"].idxmax()]

print("\n===== HIGHEST ATTENDANCE =====")
print("Student:", highest_attendance["Name"])
print("Attendance:", highest_attendance["Attendance"], "%")


# 5. Display students who scored 80 or above
high_scorers = df[df["Marks"] >= 80]

print("\n===== STUDENTS WITH MARKS 80 OR ABOVE =====")
print(high_scorers[["Name", "Marks"]])


# 6. Attendance visualization
plt.figure(figsize=(8, 5))
plt.bar(df["Name"], df["Attendance"])
plt.xlabel("Students")
plt.ylabel("Attendance (%)")
plt.title("Student Attendance Analysis")
plt.show()


# 7. Marks vs Attendance
plt.figure(figsize=(8, 5))
plt.scatter(df["Attendance"], df["Marks"])
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.title("Marks vs Attendance")
plt.show()


# 8. Final conclusion
print("\n===== CONCLUSION =====")
print("The analysis shows the marks and attendance of the students.")
print("The student with the highest marks is", highest_marks["Name"] + ".")
print("The student with the highest attendance is", highest_attendance["Name"] + ".")
print("The average marks are", round(average_marks, 2),
      "and the average attendance is", round(average_attendance, 2), "%.")
```
